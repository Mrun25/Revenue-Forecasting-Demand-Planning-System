```python
import pandas as pd
import numpy as np

from sqlalchemy import create_engine
import pymysql
```

```python
leads_df = pd.read_csv("olist_marketing_qualified_leads_dataset.csv")
deals_df = pd.read_csv("olist_closed_deals_dataset.csv")
```

```python
print(leads_df.shape)
print(deals_df.shape)
```

```python
print(leads_df.columns)
print(deals_df.columns)
```

```python
leads_df['first_contact_date'] = pd.to_datetime(
    leads_df['first_contact_date'],
    errors='coerce'
)

deals_df['won_date'] = pd.to_datetime(
    deals_df['won_date'],
    errors='coerce'
)
```

```python
print(leads_df.dtypes)
print()
print(deals_df.dtypes)
```

```python
leads_df = leads_df.drop_duplicates()
deals_df = deals_df.drop_duplicates()
```

```python
print(leads_df.isnull().sum())
print()
print(deals_df.isnull().sum())
```

```python
DB_USER = "root"
DB_PASSWORD = 1234
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "olist_db"

engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
```

```python
leads_df.to_sql(
    name="marketing_qualified_leads",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
deals_df.to_sql(
    name="closed_deals",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
print(pd.read_sql(
    "SELECT COUNT(*) AS Row_NO FROM marketing_qualified_leads",
    engine
))
print()
print(pd.read_sql(
    "SELECT COUNT(*) AS Row_NO FROM closed_deals",
    engine
))
```
